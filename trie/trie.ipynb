```python
class TrieNode:

    def __init__(self):
        self.children = [None] * 26
        self.is_end = False
    
    def _contains_char(self, ch) -> bool:
        return self.children[ord(ch) - ord('a')] is not None

    def _get_char(self, ch: str) -> 'TrieNode':
        return self.children[ord(ch) - ord('a')]

    def _put_char(self, ch: str, node: 'TrieNode') -> None:
        self.children[ord(ch) - ord('a')] = node

    def _is_end(self) -> bool:
        return self.is_end
    
    def _set_end(self) -> None:
        self.is_end = True
    

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word: str) -> None:
        node = self.root
        for w in word:
            if not node._contains_char(w):          
                node._put_char(w, TrieNode())

            node = node._get_char(w)
        node._set_end()
    
    def _search_prefix(self, word: str) -> TrieNode:
        node = self.root
        for w in word:
            if node._contains_char(w):
                node = node._get_char(w)
            else:
                return None
        return node
    
    def search(self, word: str) -> bool:
        node = self._search_prefix(word)
        return node is not None and node._is_end()

    def startsWith(self, prefix: str) -> bool:
        node = self._search_prefix(prefix)
        return True if node else False        
```

```python
import time

def run_performance_test():
    trie = Trie()

    print("Running Trie Test Suite...")

    # 1. Basic Insertion and Search
    # Scenario: Insert a standard word and verify it exists.
    trie.insert("apple")
    assert trie.search("apple") == True, "Test 1 Failed: 'apple' should be found after insertion."
    assert trie.search("app") == False, "Test 1 Failed: 'app' should not be found yet (only 'apple' exists)."

    # 2. startsWith Functionality
    # Scenario: Verify prefix detection works even if the whole word isn't marked as an end.
    assert trie.startsWith("app") == True, "Test 2 Failed: 'app' is a valid prefix of 'apple'."
    assert trie.startsWith("apple") == True, "Test 2 Failed: A word is also a prefix of itself."
    assert trie.startsWith("b") == False, "Test 2 Failed: 'b' is not in the trie."

    # 3. Inserting a Prefix of an Existing Word
    # Scenario: Insert "app" when "apple" already exists. 
    # This tests if the 'is_end' flag updates correctly on an existing node.
    trie.insert("app")
    assert trie.search("app") == True, "Test 3 Failed: 'app' should be found after explicit insertion."
    assert trie.search("apple") == True, "Test 3 Failed: 'apple' should still exist."

    # 4. Inserting a Word that Extends an Existing Word
    # Scenario: Insert "applepie" when "apple" exists.
    # This tests extending a branch deeper.
    trie.insert("applepie")
    assert trie.search("applepie") == True, "Test 4 Failed: 'applepie' should be found."
    assert trie.search("apple") == True, "Test 4 Failed: 'apple' should still be found."
    assert trie.startsWith("applep") == True, "Test 4 Failed: 'applep' is a prefix of 'applepie'."

    # 5. Branching Paths (Distinct Words)
    # Scenario: Insert "bat" and "ball". They share "ba" but split at 't' and 'l'.
    trie.insert("bat")
    trie.insert("ball")
    assert trie.search("bat") == True, "Test 5 Failed: 'bat' should be found."
    assert trie.search("ball") == True, "Test 5 Failed: 'ball' should be found."
    assert trie.startsWith("ba") == True, "Test 5 Failed: 'ba' is a prefix for both."
    assert trie.search("ba") == False, "Test 5 Failed: 'ba' was never inserted as a whole word."

    # 6. Edge Case: Empty String
    # Scenario: Inserting an empty string should mark the root as an end.
    # Based on your code logic: the loop `for w in word` won't run, `node._set_end()` runs on root.
    trie.insert("")
    assert trie.search("") == True, "Test 6 Failed: Empty string search should return True after insertion."
    assert trie.startsWith("") == True, "Test 6 Failed: Empty string is a prefix of everything/root."

    # 7. Edge Case: Single Character Word
    # Scenario: Minimal valid word length.
    trie.insert("z")
    assert trie.search("z") == True, "Test 7 Failed: Single char 'z' should be found."
    assert trie.startsWith("z") == True, "Test 7 Failed: 'z' should be a prefix."

    # 8. Redundant Insertion
    # Scenario: Inserting the same word multiple times should not change state or cause errors.
    trie.insert("cat")
    trie.insert("cat") 
    assert trie.search("cat") == True, "Test 8 Failed: 'cat' should be found after duplicate insert."

    # 9. Non-existent Prefix vs Non-existent Word
    # Scenario: Distinguish between a prefix that doesn't exist and a word that doesn't exist.
    assert trie.search("banana") == False, "Test 9 Failed: 'banana' was never inserted."
    assert trie.startsWith("ban") == False, "Test 9 Failed: No word starts with 'ban'."
    
    # 10. Long Word Stress Test
    # Scenario: Ensure recursion/iteration handles longer strings without issue.
    long_word = "a" * 100
    trie.insert(long_word)
    assert trie.search(long_word) == True, "Test 10 Failed: 100-char word should be found."
    assert trie.startsWith("a" * 50) == True, "Test 10 Failed: 50-char prefix should be valid."

    print("All 10 test cases passed successfully!")


# Start Timer
start_time = time.perf_counter()

ITERATIONS = 20
for _ in range(ITERATIONS):
    run_performance_test()

end_time = time.perf_counter()

total_time_ms = (end_time - start_time) * 1000
avg_time_ms = total_time_ms / ITERATIONS

print(f"Total time for {ITERATIONS} runs: {total_time_ms:.4f} ms")
print(f"Average time per run: {avg_time_ms:.6f} ms")
```
